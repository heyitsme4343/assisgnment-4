### Q6 
Ek Python program likhein jo kisi file mein total lines count karke bataye.

```python
# Q6 solution code here
with open("student.txt", "r") as file:
    count = 0
    line = file.readline()
    while line:
        count += 1
        line = file.readline()

print("Total lines:", count)
    
```

### Q13 (Easy)
Aisi file open karne ki koshish karein jo exist nahi karti (`non_existent_file.txt`). `FileNotFoundError` ko `try-except` se handle karein.

```python
# Q13 solution code here
try:
    with open("non_existent_file.txt", "r") as file:
        data = file.read()
        print(data)
except FileNotFoundError:
    print("Lapata chiz ko dhundney ka koi faid nhi!")
```

### Q14 
Ek program banayein jisme `finally` block ho, jo humesha execute ho chahe error aaye ya na aaye.

```python
# Q14 solution code here
try:
    number = int(input("Enter a number: "))
    print("Number:", number)
except ValueError:
    print("Ek kachori do samosa in marso ka kya bharosa")
finally:
    print("Program finished.")
```

### Q17
Ek `try-except` block likhein jo Multiple Exceptions catch kar sake (`ZeroDivisionError` aur `ValueError`).

```python
# Q17 solution code here
try:
    number = int(input("Enter a number: "))
    result = 10 / number
    print(result)

except (ZeroDivisionError, ValueError):
    print("Error: Invalid input or zero se divide nahi kar sakte!")
```

### Q18
`else` block ka istemal `try-except` ke sath karein jo sirf tab chale jab koi error na aaye.

```python
# Q18 solution code here
try:
    number = int(input("Enter a number: "))
except ValueError:
    print("Error: Please enter a valid number!")
else:
    print("Number successfully entered:", number)
```

### Q19 
`raise` keyword ka use karke age validation ka function banayein. Agar age 18 se kam ho toh custom `ValueError("Age must be 18 or older")` raise karein.

```python
# Q19 solution code here
def age_check(age):
    if age < 18:
        raise ValueError("Age must be 18 or older")
    
    print("Age is valid!")


try:
    age = int(input("Enter your age: "))
    age_check(age)

except ValueError as e:
    print("Error:", e)
```

### Q20 
File handling aur Exception handling ko combine karein: Ek file read karein, agar file na mile toh error catch karein, aur `finally` block mein file closure handle karein.

```python
# Q20 solution code here
file = None
try:
    file = open("student.txt", "r")
    data = file.read()
    print(data)

except FileNotFoundError:
    print("Error: File nahi mili!")
finally:
    if file is not None:
        file.close()
        print("File closed successfully.")
```
